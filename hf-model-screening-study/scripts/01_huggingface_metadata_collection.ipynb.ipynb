```python
from huggingface_hub import HfApi
import pandas as pd

api = HfApi()

models = api.list_models(
    pipeline_tag="text-classification",
    sort="downloads",
    limit=10
)

data = []

for model in models:
    info = api.model_info(model.id)

    data.append({
        "model_id": info.id,
        "downloads": info.downloads,
        "likes": info.likes,
        "revision": info.sha
    })

df = pd.DataFrame(data)
display(df)

df.to_csv("huggingface_models.csv", index=False)
```

```python
from pathlib import PurePosixPath

for model in data:
    info = api.model_info(model["model_id"])

    files = [f.rfilename for f in info.siblings]

    model["file_count"] = len(files)
    model["has_safetensors"] = any(
        f.endswith(".safetensors") for f in files
    )
    model["has_pickle_like_files"] = any(
        PurePosixPath(f).suffix.lower() in
        {".bin", ".pt", ".pth", ".pkl", ".pickle"}
        for f in files
    )
    model["has_python_files"] = any(
        f.endswith(".py") for f in files
    )

df = pd.DataFrame(data)
display(df)

df.to_csv("huggingface_models.csv", index=False)
```
